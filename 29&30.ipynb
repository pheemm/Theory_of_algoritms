# 29

```python
from bisect import bisect_left

n = int(input())

points = []

for _ in range(n):
    x, r = map(int, input().split())
    points.append((x, r))

points.sort()

x = [0] + [point[0] for point in points]
r = [0] + [point[1] for point in points]

dp = [0] * (n + 1)

for i in range(1, n + 1):
    limit = x[i] - 5
    j = bisect_left(x, limit) - 1

    dp[i] = max(dp[i - 1], r[i] + dp[j])

print(dp[n])
```

```python
n = int(input())

w = list(map(int, input().split()))

dp = [0] * (n + 1)

dp[1] = w[0]

for i in range(2, n + 1):
    dp[i] = max(dp[i - 1], dp[i - 2] + w[i - 1])

print(dp[n])
```
